### Task 1: Load, Clean, and Normalize Weather Features
We will start by importing the necessary libraries and loading the train and test CSV files.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# Load datasets
train_df = pd.read_csv('/content/DailyDelhiClimateTrain.csv')
test_df = pd.read_csv('/content/DailyDelhiClimateTest.csv')

print("Train dataset shape:", train_df.shape)
print("Test dataset shape:", test_df.shape)

print("\nTrain Info:")
train_df.info()

print("\nFirst 5 rows of Train:")
display(train_df.head())
```

Next, let's parse the date column, handle any missing values, and scale/normalize the features.

```python
# Convert date column to datetime
train_df['date'] = pd.to_datetime(train_df['date'])
test_df['date'] = pd.to_datetime(test_df['date'])

train_df.set_index('date', inplace=True)
test_df.set_index('date', inplace=True)

# Check for missing values
print("Missing values in Train:")
print(train_df.isnull().sum())
print("\nMissing values in Test:")
print(test_df.isnull().sum())

# Normalize the weather features
scaler = MinMaxScaler()

# We will fit the scaler on the train set and transform both train and test sets to prevent data leakage
scaled_train = pd.DataFrame(scaler.fit_transform(train_df), columns=train_df.columns, index=train_df.index)
scaled_test = pd.DataFrame(scaler.transform(test_df), columns=test_df.columns, index=test_df.index)

print("\nNormalized Train head:")
display(scaled_train.head())
```

### Task 2: Generate sequences
We will define a sequence generator function. We will use a sliding window of past weather measurements (e.g., 30 days) of all features (`meantemp`, `humidity`, `wind_speed`, `meanpressure`) to forecast the target variable: the next day's `meantemp` (first column).

```python
def create_sequences(data, seq_length=30):
    X = []
    y = []
    # We want to forecast the 'meantemp' which is at index 0
    for i in range(len(data) - seq_length):
        X.append(data.iloc[i:i+seq_length].values)
        y.append(data.iloc[i+seq_length, 0])
    return np.array(X), np.array(y)

SEQUENCE_LENGTH = 30

X_train, y_train = create_sequences(scaled_train, SEQUENCE_LENGTH)

# For the test set, we need to prepend the last SEQUENCE_LENGTH days of train set
# so that we can forecast starting from the very first day of the test set
combined_test_data = pd.concat([scaled_train.iloc[-SEQUENCE_LENGTH:], scaled_test])
X_test, y_test = create_sequences(combined_test_data, SEQUENCE_LENGTH)

print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")
```

### Task 3: Build and Train the Sequence Model (LSTM)
Now we build the LSTM architecture using TensorFlow/Keras and train it.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

tf.random.set_seed(42)
np.random.seed(42)

model = Sequential([
    LSTM(64, activation='tanh', return_sequences=True, input_shape=(SEQUENCE_LENGTH, X_train.shape[2])),
    Dropout(0.1),
    LSTM(32, activation='tanh'),
    Dropout(0.1),
    Dense(16, activation='relu'),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.summary()

early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

history = model.fit(
    X_train, y_train,
    epochs=60,
    batch_size=32,
    validation_split=0.15,
    callbacks=[early_stop],
    verbose=1
)
```

### Task 4: Evaluate and Plot Actual vs Predicted Temperature
We will use our trained model to forecast the temperature on the test set, evaluate the forecasts using Mean Absolute Error (MAE) and Root Mean Squared Error (RMSE), and plot the comparisons.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 1. Predict on the test set
predictions_scaled = model.predict(X_test)

# To inverse-transform, we need a dummy array matching the scaler's feature size (4 features)
dummy_pred = np.zeros((len(predictions_scaled), 4))
dummy_pred[:, 0] = predictions_scaled.flatten() # place predicted meantemp in first column

inversed_pred = scaler.inverse_transform(dummy_pred)[:, 0]
actual_temp = test_df['meantemp'].values

# 2. Calculate Evaluation Metrics
mae = mean_absolute_error(actual_temp, inversed_pred)
rmse = np.sqrt(mean_squared_error(actual_temp, inversed_pred))

print(f"\n--- Evaluation Metrics ---")
print(f"Mean Absolute Error (MAE): {mae:.4f} °C")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f} °C")
```
